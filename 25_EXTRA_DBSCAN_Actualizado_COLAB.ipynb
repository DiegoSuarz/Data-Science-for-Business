#### 3.4 Outliers

```python
# Cálculo de intervalo del diagrama de cajas - Método de Rango Intercuartílico
#def calculateNumOutliars(serie):
#  Q01 = serie.quantile(0.25)
#  Q03 = serie.quantile(0.75)
#  IQR = Q03 - Q01
#  a = (serie < (Q01 - 1.5 * IQR)) | (serie > (Q03 + 1.5 * IQR))
#  numOutliars = a[a == True].shape[0]
#  return numOutliars
```

```python
# Usamos el método de Z-score (considerando se distribuye Normalmente) --- para grandes volúmenes de datos
def calculateNumOutliars(serie):
    mu = serie.mean()
    desv = np.std(serie)
    a = ((serie-mu)/desv < -2) | ((serie-mu)/desv > 2)
    numOutliars = a[a == True].shape[0]
    return a,numOutliars    
```

```python
numTotal = dataFrame.shape[0]
for columnName in rubroName:
    a,numOutliars = calculateNumOutliars(dataFrame[columnName])
    # Creamos nuevos campos para filtrar los Outliers 
    dataFrame['flg_'+columnName]=a
    print('*'+columnName)
    if numOutliars > 0:
      print("Número de valores outliars: " + str(numOutliars))
      print("Porcentaje: " + str(np.round(numOutliars * 100 / numTotal, 2)) + "%")
    else:
      print("****No hay Outliers")    
    print("\n")
```

#### 3.5 Reducción de dimensión (PCA)

```python
# Calculamos el máximo número de componentes (Nro variables = Nro máximo de componentes)
from sklearn.decomposition import PCA

pca = PCA()
pca.fit(dataFrame[rubroName])
pca.explained_variance_ratio_
```

```python
# Imprimir porcentaje de varianza explicada por cada componente
for i in range(len(pca.components_)):
    print('% Var. explicada ('+str(i+1)+' componentes): ', np.cumsum(pca.explained_variance_ratio_)[i]*100)

# Graficar la varianza explicada por cada componente
plt.bar(range(1, len(pca.components_)+1), pca.explained_variance_ratio_, alpha=0.5, color='gray')
plt.plot(range(1, len(pca.components_)+1), np.cumsum(pca.explained_variance_ratio_), alpha=0.7, color='red', marker='o')

# Títulos y etiquetas
plt.title("Varianza explicada y curva de pareto")
plt.xlabel('Número de componentes principales')
plt.ylabel('Porcentaje de varianza explicada')

# Mostrar el gráfico
plt.show()
```

#### 3.6 Modelamiento

```python
from sklearn.cluster import DBSCAN
from sklearn import metrics
from sklearn.metrics import pairwise_distances_argmin_min 
```

```python
# Calculando el número de clúster adecuado:
X = dataFrame[rubroNameComp]

# Número de puntos mínimos (muestra)
numSam = range(4, 15)
dbscan = [DBSCAN(eps=0.25, min_samples=i) for i in numSam] #epsion de 0.25
dbscan
modDbscan = [dbscan[i].fit(X) for i in range(len(dbscan))]
modDbscan
cluster = [modDbscan[i].labels_ for i in range(len(dbscan))] 
silhou  = [metrics.silhouette_score(X, cluster[i]) for i in range(len(dbscan))] 

plt.plot(numSam,silhou)
plt.xlabel('Muestra')
plt.ylabel('Silueta')
plt.title('Parámetro DBSCAN')
plt.show()
```

Segun el grafico, la cantidad optima de vecinos esta entre 8 y 10

```python
# Nos fijamos de los indicadores de Clustering:

for i in range(len(numSam)):
    print('* Posición Nro: '+str(i+1))
    print(str(numSam[i])+' puntos mínimos:')
    print('Ctd clúster: '+str(len(set(cluster[i]))))
    print('Silueta: '+str(silhou[i]))
    print("\n")
```

```python
dfTransp = centroide.T
corr = dfTransp.corr()

def plot_correlations(corr):
    sns.set(style="white")
    
    # Crear un mapa de colores divergente
    cmap = sns.diverging_palette(220, 10, as_cmap=True)
    
    # Crear la máscara para el triángulo superior
    mask = np.zeros_like(corr, dtype=bool)  # Cambiado de np.bool a bool
    mask[np.triu_indices_from(mask)] = True
    
    # Dibujar el mapa de calor con la máscara y el formato adecuado
    sns.heatmap(abs(corr), mask=mask, cmap=cmap, vmax=1, center=0, square=True, 
                linewidths=.5, cbar_kws={"shrink": .5})

# Llamar a la función para graficar
plot_correlations(corr)
```

```python
print('  ******************************')
print('       GRUPOS DEL CLUSTER '+str(len(set(clusFinal['clusterFin']))))
print('  ******************************')

# Tener en cuenta el clúster con etiqueta "-1" que son los anómalos
indx = clusFinal.groupby(clusFinal['clusterFin']).count().index

for i in indx:
    if i not in centroide.index:
        print(f"El índice {i} no se encuentra en el dataframe centroide.")
        continue
    
    mu = centroide.loc[i].mean()
    dev = np.std(centroide.loc[i])
    
    # Verificación de divisiones por cero
    if dev == 0:
        print(f"Desviación estándar cero para el índice {i}, omitiendo...")
        continue
    
    stats = (centroide.loc[i].values - mu) / dev
   
    # Asegurarse de que rubroNameComp tiene el mismo número de elementos que stats
    if len(rubroNameComp) != len(stats):
        print(f"Error: La longitud de rubroNameComp ({len(rubroNameComp)}) no coincide con stats ({len(stats)}).")
        continue

    angles = np.linspace(0, 2 * np.pi, len(rubroNameComp), endpoint=False)
    stats = np.concatenate((stats, [stats[0]]))  # Cerrar el gráfico
    ref = [0] * (len(rubroNameComp) + 1)         # Referencia en cero
    angles = np.concatenate((angles, [angles[0]]))  # Cerrar los ángulos
    
    # Crear gráfico polar
    fig = plt.figure()
    ax = fig.add_subplot(111, polar=True)
    
    # Graficar los valores de stats
    ax.plot(angles, stats, 'o-', linewidth=3, label=f"Grupo {i}")
    ax.plot(angles, ref, '-', linewidth=2)
    ax.fill(angles, stats, alpha=0.25)
    
    # Configurar las etiquetas y el grid
    #ax.set_thetagrids(angles * 180 / np.pi, rubroNameComp)
    # Verificar que el tamaño de angles coincide con el de rubroNameComp
    if len(angles) - 1 != len(rubroNameComp):  # angles tiene un elemento más por el cierre
        print(f"Error: La longitud de rubroNameComp ({len(rubroNameComp)}) no coincide con el número de ángulos ({len(angles) - 1}).")
    else:
        ax.set_thetagrids(angles[:-1] * 180 / np.pi, rubroNameComp)
    ax.grid(True)
    ax.set_title(f'Grupo {i}\n')
    
    plt.legend()
    plt.show()
```
